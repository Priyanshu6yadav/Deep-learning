```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# Sentences
sent =[ 'the glass of milk',
       'the glass of juice',
       'the cup of tea',
       'I am a good boy',
       'I am a good developer',
       'understand the meaning of words',
       'your video are good',
]
```

```python
# Define the vocabulary size
voc_size = 10000
```

```python
sent
```

```python
## One hot representantion
one_hot_repr = [one_hot(words,voc_size)for words in sent]
one_hot_repr
```

```python
# Word Embedding Representation
from tensorflow.keras.layers import Embedding
# from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
# Pad sequences to a fixed length using pre-padding
sent_length = 8   # longest sentence has 5 words; 8 is just a chosen fixed length
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
print(embedded_docs)
```

```python
## Feature representation
dim = 10 # Dimension
```

```python
from tensorflow.keras.layers import Input, Embedding
from tensorflow.keras.models import Sequential

model = Sequential([
    Input(shape=(sent_length,)),
    Embedding(input_dim=voc_size, output_dim=dim),
])
model.compile(optimizer='adam', loss='mse')
model.summary()
```

```python
model.predict(embedded_docs) # Every words will be represented by 10 dimension
```
